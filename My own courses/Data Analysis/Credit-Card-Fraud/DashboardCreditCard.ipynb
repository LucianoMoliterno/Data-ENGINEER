```python
# Importar bibliotecas
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display
```

```python
# Cargar los datos
df = pd.read_csv('creditcard.csv')

# Preprocesamiento
X = df.drop(columns=['Class'])
y = df['Class']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Entrenar modelo
model = RandomForestClassifier(random_state=42)
model.fit(X_scaled, y)
```

```python
# Widget para ingresar datos
input_widgets = []
for i in range(X.shape[1]):
    widget = widgets.FloatText(description=f'Feature {i+1}', value=0.0)
    input_widgets.append(widget)

# Botón para hacer predicciones
predict_button = widgets.Button(description="Predecir")

# Salida para mostrar el resultado
output = widgets.Output()

# Función para manejar la predicción
def on_predict_button_clicked(b):
    with output:
        output.clear_output()
        input_values = [w.value for w in input_widgets]
        input_array = np.array(input_values).reshape(1, -1)
        scaled_input = scaler.transform(input_array)
        prediction = model.predict(scaled_input)
        result = "Fraude" if prediction[0] == 1 else "No Fraude"
        print(f"Resultado: {result}")

predict_button.on_click(on_predict_button_clicked)

# Mostrar widgets
display(*input_widgets, predict_button, output)
```

```python
# Histograma interactivo del monto de transacciones
fig = px.histogram(df, x='Amount', color='Class', nbins=50,
                   title='Distribución del Monto de Transacciones por Clase',
                   labels={'Amount': 'Monto', 'Class': 'Clase'})
fig.show()
# Gráfico de dispersión interactivo
fig = px.scatter(df, x='V1', y='V2', color='Class',
                 title='Gráfico de Dispersión de V1 vs V2',
                 labels={'V1': 'Variable V1', 'V2': 'Variable V2', 'Class': 'Clase'})
fig.show()
```
